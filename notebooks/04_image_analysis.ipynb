```python
from pathlib import Path
import sys
PROJECT_ROOT = Path.cwd().parent
sys.path.append(str(PROJECT_ROOT))

import matplotlib.pyplot as plt
from src.analysis.image_analysis import analyze_image_properties
```

```python
image_dir = Path(
    "../data/raw/rsna-pneumonia-detection-challenge/stage_2_train_images"
)

image_df = analyze_image_properties(image_dir)
```

```python
image_df.head()
```

```python
# Image Width Distribution

plt.figure(figsize=(8,5))

plt.hist(
    image_df["width"],
    bins=20
)

plt.title("Image Width Distribution")

plt.xlabel("Pixels")

plt.ylabel("Images")

plt.show()
```

```python
# Image Height Distribution

plt.figure(figsize=(8,5))

plt.hist(
    image_df["height"],
    bins=20
)

plt.title("Image Height Distribution")

plt.xlabel("Pixels")

plt.ylabel("Images")

plt.show()
```

```python
# Aspect Ratio
plt.figure(figsize=(8,5))

plt.hist(
    image_df["aspect_ratio"],
    bins=20
)

plt.title("Aspect Ratio Distribution")

plt.show()
```

```python
# Pixel Spacing
image_df["pixel_spacing_x"].plot(
    kind="hist",
    bins=20
)

plt.title("Pixel Spacing (X)")

plt.show()

# Repeat for Y if needed.
```

```python
# View Position

image_df["view_position"].value_counts().plot(
    kind="bar"
)

plt.title("View Position Distribution")

plt.xlabel("View Position")

plt.ylabel("Images")

plt.show()
```

```python
# Width Boxplot

plt.boxplot(image_df["width"])

plt.title("Width Boxplot")

plt.show()
```

```python
# Height Boxplot

plt.boxplot(image_df["height"])

plt.title("Height Boxplot")

plt.show()
```

```python
# Aspect Ratio Boxplot

plt.boxplot(image_df["aspect_ratio"])

plt.title("Aspect Ratio Boxplot")

plt.show()
```
